## Day 15: Unsupervised Learning & User Segmentation

```python
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
import numpy as np

X, _ = make_blobs(n_samples=300, centers=5, cluster_std=0.60, random_state=0)

# Scale the data 
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Run 1: Random Init
kmeans_random1 = KMeans(n_clusters=5, init='random', random_state=42, n_init=10)
kmeans_random1.fit(X_scaled)

# Run 2: Random Init (different seed)
kmeans_random2 = KMeans(n_clusters=5, init='random', random_state=7, n_init=10)
kmeans_random2.fit(X_scaled)

# Run 3: K-Means++
kmeans_plus = KMeans(n_clusters=5, init='k-means++', random_state=42, n_init=10)
kmeans_plus.fit(X_scaled)

# Store centers
centers_1 = kmeans_random1.cluster_centers_
centers_2 = kmeans_random2.cluster_centers_
centers_3 = kmeans_plus.cluster_centers_

print("Random Init (Run 1):\n", centers_1)
print("\nRandom Init (Run 2):\n", centers_2)
print("\nK-Means++:\n", centers_3)
```

```python
# Measure difference between cluster centers
diff_1_2 = np.linalg.norm(centers_1 - centers_2)
diff_1_3 = np.linalg.norm(centers_1 - centers_3)

print("Difference (Random1 vs Random2):", diff_1_2)
print("Difference (Random1 vs KMeans++):", diff_1_3)
```

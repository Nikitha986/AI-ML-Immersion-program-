## 🔹 What is Vectorization?

Vectorization means performing operations on an entire dataset at once instead of using loops.

### Traditional Python:
- Processes one element at a time using loops

### NumPy:
- Processes all elements together
- Uses optimized low-level implementation

### Why it matters:
- Faster execution
- Efficient for large datasets
- Used in AI for images, text, and numerical data

```python
# Example without vectorization (loop)
data = [10, 20, 30, 40]
result = []

for i in data:
    result.append(i * 2)

print("Using loop:", result)
```

```python
# Example with vectorization (NumPy)
import numpy as np

data = np.array([10, 20, 30, 40])
result = data * 2

print("Using NumPy:", result)
```

## 🔹 NumPy Reshaping

In AI, data is often represented as matrices.

For example:
- Images → pixel grids
- Text → numerical embeddings

Reshaping allows us to convert data into formats suitable for models.

```python
data_points = np.array([10, 20, 30, 40])

matrix = data_points.reshape(2, 2)

print("2x2 Matrix:\n", matrix)
```

## 🔹 Key Learning

- Vectorization improves performance by avoiding loops
- NumPy allows efficient numerical computation
- Data can be reshaped into matrices for AI models

This concept is essential for handling large-scale datasets in machine learning.

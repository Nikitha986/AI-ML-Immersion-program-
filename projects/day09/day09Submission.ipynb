## Day 9: Model Optimization & Grid Search

```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
# 1. Load and Scale (Same as Day 8)
data = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
# 2. Define the "Grid" of parameters to try
# We want to test different values of 'alpha'
param_grid = {'alpha': [0.1, 1.0, 10.0, 100.0, 500.0]}
# 3. Initialize Grid Search
# cv=5 means 5-fold Cross-Validation (splitting data 5 times to be sure)
grid_search = GridSearchCV(Ridge(), param_grid, cv=5, scoring='r2')
# 4. RUN THE SEARCH
grid_search.fit(X_train_scaled, y_train)

print(f"Best Alpha Found: {grid_search.best_params_}")
print(f"Best R2 Score: {grid_search.best_score_:.4f}")
```

## Reflection: Why start with a wider range of values?

When beginning a Grid Search, it is better to test a **wide range of values** (e.g., [0.1, 1, 10, 100]) rather than small increments (e.g., [1.1, 1.2, 1.3]). A wide search helps us quickly identify the correct *scale* of the hyperparameter. If the best value is far from our initial guess, small increments would miss it entirely. Once the right scale is found, we can refine with smaller steps around that region. This strategy saves time and ensures we don’t overlook the true optimal setting.


```python
import pandas as pd
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler

# 1. Load and Scale
data = fetch_california_housing()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.2, random_state=42)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 2. Define parameter grid
param_grid = {'alpha': [0.1, 1.0, 10.0, 100.0, 500.0]}

# 3. Grid Search
grid_search = GridSearchCV(Ridge(), param_grid, cv=5, scoring='r2')
grid_search.fit(X_train_scaled, y_train)

# 4. Compare with default Ridge
ridge_default = Ridge()
ridge_default.fit(X_train_scaled, y_train)

print(f"Best Alpha Found: {grid_search.best_params_['alpha']}")
print(f"Best R2 Score (CV): {grid_search.best_score_:.4f}")

```
